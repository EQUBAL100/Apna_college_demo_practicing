Convert Binary OVF → ASCII (text OVF2 format)

Processing of ovf file to generate bits

```
import numpy as np
import glob
import re

# =========================
# READ OVF + GET GRID SIZE
# =========================
def read_ovf(filename):

    with open(filename, 'rb') as f:
        content = f.read()

    try:
        # ---------- ASCII ----------
        text = content.decode('utf-8')
        lines = text.split('\n')

        Nx = Ny = None

        for i, line in enumerate(lines):
            if "xnodes" in line:
                Nx = int(re.findall(r'\d+', line)[0])
            if "ynodes" in line:
                Ny = int(re.findall(r'\d+', line)[0])
            if "Begin: Data" in line:
                start = i + 1
                break

        data = np.loadtxt(lines[start:])

    except:
        # ---------- BINARY ----------
        with open(filename, 'rb') as f:
            Nx = Ny = None

            while True:
                line = f.readline()

                if b'xnodes' in line:
                    Nx = int(re.findall(rb'\d+', line)[0])
                if b'ynodes' in line:
                    Ny = int(re.findall(rb'\d+', line)[0])

                if b'Begin: Data' in line:
                    break

            # Skip control number
            f.read(4)

            data = np.fromfile(f, dtype=np.float32)

        # reshape properly
        N = Nx * Ny
        data = data[:3*N].reshape((N, 3))

    mx = data[:, 0]
    my = data[:, 1]
    mz = data[:, 2]

    return mx, my, mz, Nx, Ny


# =========================
# RESHAPE
# =========================
def reshape_field(mx, my, mz, Nx, Ny):
    mx = mx.reshape(Ny, Nx)
    my = my.reshape(Ny, Nx)
    mz = mz.reshape(Ny, Nx)
    return mx, my, mz


# =========================
# POLARITY
# =========================
def compute_polarity(mz):
    Ny, Nx = mz.shape
    cx, cy = Nx//2, Ny//2

    core = mz[cy-2:cy+3, cx-2:cx+3]
    return 1 if np.mean(core) > 0 else 0


# =========================
# CHIRALITY (STABLE)
# =========================
def compute_chirality(mx, my):
    Ny, Nx = mx.shape
    cx, cy = Nx//2, Ny//2

    r = min(cx, cy) // 2
    n = 100

    circulation = 0.0

    for k in range(n):
        theta = 2*np.pi*k/n

        x = int(cx + r*np.cos(theta))
        y = int(cy + r*np.sin(theta))

        if 0 <= x < Nx and 0 <= y < Ny:
            mx_val = mx[y, x]
            my_val = my[y, x]

            tx = -np.sin(theta)
            ty =  np.cos(theta)

            circulation += mx_val*tx + my_val*ty

    return 1 if circulation > 0 else 0


# =========================
# MAIN
# =========================
files = sorted(glob.glob("Vortex_TRNG.out/m*.ovf"))

bitstream = []

for f in files:
    mx, my, mz, Nx, Ny = read_ovf(f)

    mx, my, mz = reshape_field(mx, my, mz, Nx, Ny)

    p = compute_polarity(mz)
    c = compute_chirality(mx, my)

    bitstream.append(p)
    bitstream.append(c)

bitstring = ''.join(map(str, bitstream))

with open("bitstream.txt", "w") as f:
    f.write(bitstring)

print("✔ Bitstream generated")
print("Total bits:", len(bitstring))
print("First 100 bits:", bitstring[:100])
```

```
# =====================================
# GENERATE BITSTREAM FILES
# =====================================

# bitstream already contains:
# [p1, c1, p2, c2, ...]

# =====================================
# 1. RAW TXT (NO COMMAS)
# Example:
# 011001010101...
# =====================================

bitstring = ''.join(map(str, bitstream))

with open("bitstream.txt", "w") as f:
    f.write(bitstring)

print("✔ bitstream.txt saved")


# =====================================
# 2. COMMA-SEPARATED TXT
# Example:
# 0,1,1,0,0,1,...
# =====================================

bitstring_comma = ','.join(map(str, bitstream))

with open("bitstream_comma.txt", "w") as f:
    f.write(bitstring_comma)

print("✔ bitstream_comma.txt saved")


# =====================================
# 3. BINARY FILE (.bin)
# NIST-ready packed binary
# =====================================

import numpy as np

bits_array = np.array(bitstream, dtype=np.uint8)

# convert bits → bytes
packed_bits = np.packbits(bits_array)

with open("bitstream.bin", "wb") as f:
    f.write(packed_bits)

print("✔ bitstream.bin saved")


# =====================================
# 4. SUMMARY
# =====================================

print("\nTotal bits =", len(bitstream))

print("\nFirst 50 bits (raw):")
print(bitstring[:50])

print("\nFirst 20 comma-separated bits:")
print(bitstring_comma.split(',')[:20])
```
